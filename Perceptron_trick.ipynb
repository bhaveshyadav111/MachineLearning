```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
```

```python
X,y = make_classification(n_samples=600,n_features=2,n_informative=1,n_redundant=0,n_classes=2,n_clusters_per_class=1,random_state=1)
```

```python
X
```

```python
sns.scatterplot(x=X[:,0],y=X[:,1],hue=y)
```

```python
def perceptron(X,y) :
  X = np.insert(X,0,1,axis=1)
  w = np.ones(X.shape[1])
  lr = 0.1

  for epoch in range(1000) :
    nth = np.random.randint(X.shape[0])

    net = np.dot(X[nth],w)
    ypred = sign(net)
    w = w + lr *(y[nth]-ypred) * X[nth]

  return w[0],w[1:]


def sign(net):
  return np.where(net > 0,1,0)
```

```python
intercept_,coef_ = perceptron(X,y)
print(intercept_)
print(coef_)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
print(m,b)
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + b
```

```python
sns.scatterplot(x=X[:,0],y=X[:,1],hue=y)
plt.plot(x_input,y_input,color='red')
```
